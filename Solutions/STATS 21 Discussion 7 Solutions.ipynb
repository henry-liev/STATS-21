### 1. Matplotlib: Scatter Plot with Reference Line
Write a function `plot_exam_scatter(df, passing_score=75)` that:
* creates a figure using `fig, ax = plt.subplots(figsize=(7, 4.5))`
* plots `'hours_studied'` vs `'exam_score'` using `ax.scatter()` with `color='steelblue'`, `edgecolor='black'`, `s=70`, `alpha=0.85`, and `label='Students'`
* adds a horizontal line at `passing_score` using `ax.axhline()` with `color='crimson'`, `linestyle='--'`, and `label=f'Passing Score ({passing_score})'`
* sets the title to `'Hours Studied vs. Exam Score'`, x-axis label to `'Hours Studied'`, and y-axis label to `'Exam Score'`
* sets y-axis limits to `(50, 105)` and adds a grid with `linestyle=':'` and `alpha=0.6`
* adds a legend, calls `fig.tight_layout()`, and returns `(fig, ax)`


```python
df_students = pd.DataFrame({
    'student': ['Alice', 'Bob', 'Charlie', 'Diana', 'Evan', 'Fiona', 'George', 'Hannah'],
    'hours_studied': [2.5, 5.0, 7.5, 3.0, 8.5, 6.0, 1.5, 9.0],
    'exam_score': [68, 82, 91, 74, 95, 86, 58, 98]
})

def plot_exam_scatter(df, passing_score=75):
    fig, ax = plt.subplots(figsize=(7, 4.5))
    ax.scatter(df['hours_studied'], df['exam_score'], color='steelblue', edgecolor='black', s=70, alpha=0.85, label='Students')
    ax.axhline(passing_score, color='crimson', linestyle='--', linewidth=1.5, label=f'Passing Score ({passing_score})')
    ax.set_title('Hours Studied vs. Exam Score')
    ax.set_xlabel('Hours Studied')
    ax.set_ylabel('Exam Score')
    ax.set_ylim(50, 105)
    ax.grid(True, linestyle=':', alpha=0.6)
    ax.legend()
    fig.tight_layout()
    return fig, ax
```

```python
fig, ax = plot_exam_scatter(df_students)
plt.show()
```

### 2. Seaborn: Grouped Bar Plot
Write a function `plot_penguin_mass(df)` that:
* creates a figure using `fig, ax = plt.subplots(figsize=(7, 4.5))`
* uses `sns.barplot()` to plot average `'body_mass_g'` by `'species'` with `hue='sex'`, `palette='Set2'`, `errorbar=None`, and `ax=ax`
* sets the title to `'Average Penguin Body Mass by Species and Sex'`
* sets the x-axis label to `'Species'` and y-axis label to `'Average Body Mass (g)'`
* calls `fig.tight_layout()` and returns `(fig, ax)`


```python
df_penguins = pd.DataFrame({
    'species': ['Adelie', 'Adelie', 'Adelie', 'Adelie', 'Adelie', 'Adelie',
                'Chinstrap', 'Chinstrap', 'Chinstrap', 'Chinstrap', 'Chinstrap', 'Chinstrap',
                'Gentoo', 'Gentoo', 'Gentoo', 'Gentoo', 'Gentoo', 'Gentoo'],
    'island': ['Torgersen', 'Torgersen', 'Biscoe', 'Biscoe', 'Dream', 'Dream',
               'Dream', 'Dream', 'Dream', 'Dream', 'Dream', 'Dream',
               'Biscoe', 'Biscoe', 'Biscoe', 'Biscoe', 'Biscoe', 'Biscoe'],
    'sex': ['male', 'female', 'male', 'female', 'male', 'female',
            'male', 'female', 'male', 'female', 'male', 'female',
            'male', 'female', 'male', 'female', 'male', 'female'],
    'body_mass_g': [3750, 3450, 3900, 3625, 3950, 3150,
                    3500, 3200, 4050, 3650, 4300, 3450,
                    4500, 4750, 5200, 4650, 5700, 4850]
})

def plot_penguin_mass(df):
    fig, ax = plt.subplots(figsize=(7, 4.5))
    sns.barplot(data=df, x='species', y='body_mass_g', hue='sex', palette='Set2', errorbar=None, ax=ax)
    ax.set_title('Average Penguin Body Mass by Species and Sex')
    ax.set_xlabel('Species')
    ax.set_ylabel('Average Body Mass (g)')
    fig.tight_layout()
    return fig, ax
```

```python
fig, ax = plot_penguin_mass(df_penguins)
plt.show()
```
